```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

df = pd.read_csv("final_dependency_risk_scores.csv")
```

```python
sns.set(style="whitegrid")
```

```python
plt.figure(figsize=(12, 6))
sns.barplot(data=df, x="Dependency", y="Final Risk Score", hue="Risk Level", dodge=False)
plt.title("Final Risk Score per Dependency")
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.savefig("plot_final_risk_scores_bar.png")
plt.show()
```

```python
plt.figure(figsize=(6, 4))
df['Risk Level'].value_counts().plot.pie(autopct='%1.1f%%', colors=['red', 'orange', 'green'])
plt.title("Risk Level Distribution")
plt.ylabel("")
plt.savefig("plot_risk_level_pie.png")
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load CSV
df = pd.read_csv("final_dependency_risk_scores.csv")

# Ensure only relevant columns exist
# Assuming columns: Project, Dependency, Final Score
# Pivot to create matrix: rows = projects, columns = dependencies
heatmap_data = df.pivot(index="Project", columns="Dependency", values="Final Score")

# Plot heatmap
plt.figure(figsize=(12, 8))
sns.heatmap(heatmap_data, annot=True, fmt=".1f", cmap="coolwarm", cbar_kws={'label': 'Final Risk Score'})
plt.title("Heatmap of Final Risk Scores Across Projects & Dependencies", fontsize=14)
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```
