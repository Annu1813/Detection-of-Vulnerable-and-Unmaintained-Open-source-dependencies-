```python
import pandas as pd
from sklearn.metrics import confusion_matrix, classification_report
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("phase4A_ground_truth_labeled.csv")
```

```python
df_filtered = df[df["ground_truth"].isin(["Abandoned", "Active"])]

df_filtered["predicted_label"] = df_filtered["ARS"].apply(lambda x: "Abandoned" if x >= 6 else "Active")
```

```python
y_true = df_filtered["ground_truth"]
y_pred = df_filtered["predicted_label"]

labels = ["Abandoned", "Active"]
cm = confusion_matrix(y_true, y_pred, labels=labels)
```

```python
print("Confusion Matrix:\n", cm)

plt.figure(figsize=(6, 5))
sns.heatmap(cm, annot=True, fmt="d", cmap="Blues", xticklabels=labels, yticklabels=labels)
plt.title("Confusion Matrix")
plt.xlabel("Predicted Label")
plt.ylabel("True Label")
plt.tight_layout()
plt.show()
```

```python
cm = confusion_matrix(y_true, y_pred, labels=["Abandoned", "Active"])
print("Confusion Matrix:\n", cm)
```

```python
report = classification_report(y_true, y_pred, digits=2)
print("\nClassification Report:\n", report)
```
