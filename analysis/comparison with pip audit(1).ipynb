```python
import pandas as pd

# Load your final scores CSV
df = pd.read_csv("final_dependency_risk_scores.csv")

# Select key comparison columns
comparison_table = df[[
    "Dependency",             # Package name
    "CVSS",                   # pip-audit CVSS Score
    "ARS",                    # Abandonment Risk Score
    "Final Risk Score",       # Combined score
    "Risk Level"              # Classified Risk Level
]]

# Sort by Final Risk Score descending (optional)
comparison_table = comparison_table.sort_values(by="Final Risk Score", ascending=False)

# Save table to CSV for documentation or report
comparison_table.to_csv("comparison_with_pip_audit.csv", index=False)

# Show top 10 rows
print(comparison_table.head(10))
```
